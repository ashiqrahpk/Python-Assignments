```python
#2. Find the performance category based on the input score using following criteria:
#a. Above Average:  Score greater than 7
#b. Average: Score between 4 and 7(both inclusive)
#c. Below Average: Score lesser than 4
num = int(input("Enter a number from 0 to 10: "))
if num < 0 or num > 10:
  print("number not possible")
elif num > 7:
  print("Above Average")
elif num >= 4 and num <= 7:
  print("Average")
else:
  print("Below Average")
```

```python
#3. Output: Print the Performance category
num = int(input("Enter a number from 0 to 10: "))
if num < 0 or num > 10:
  print("number not possible")
elif num > 7:
  print("The performance category is:","Above Average")
elif num >= 4 and num <= 7:
  print("The performance category is:","Average")
else:
  print("The performance category is:","Below Average")
```
